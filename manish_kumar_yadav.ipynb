```python
#task 4
# Initialize an empty list to store inputs
strings_list = []

# Continuously prompt for input until an empty string is entered
while True:
    user_input = input("Enter a string (or press Enter to finish): ")
    if user_input == "":
        break
    strings_list.append(user_input)

# Sort the list alphabetically
strings_list.sort()

# Display the strings in lower case and alphabetical order
print("\nSorted strings in lower case:")
for s in strings_list:
    print(s.lower())
```
